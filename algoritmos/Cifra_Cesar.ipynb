# Cifra Cesar

```python
def cifra_cesar(texto, chave):
    resultado = ""
    for i in range(len(texto)):
        letra = texto[i]
        # Verifica se o caractere é uma letra, ignorando números e caracteres especiais.
        if letra.isalpha():
            # Verifica se o caractere é uma letra maiúscula
            if (letra.isupper()):
                resultado += chr((ord(letra) + chave-65) % 26 + 65)
            # Verifica se o caractere é uma letra minúscula
            elif (letra.islower()):
                resultado += chr((ord(letra) + chave-97) % 26 + 97)
        else:
            resultado += letra

    return resultado

def exibe_exemplo(texto, chave):
    cifra = cifra_cesar(texto, chave)
    print(f"Texto Original: {texto}")
    print(f"Texto Cifrado: {cifra} (Chave={chave})")
    return
```

```python
texto = "Semana de Tecnologia"

for chave in [3, 8, 13]:
    exibe_exemplo(texto, chave)
    print("-"*10)
```

```python
chave = 10
cifra = cifra_cesar(texto, chave)
print("Rotação | Texto")
for i in range(26):
    nova_cifra = cifra_cesar(cifra, i)
    print(f"{i} | {nova_cifra}")
```
